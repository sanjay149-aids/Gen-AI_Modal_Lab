```python
# Input list
colors = ["red", "blue", "green", "red", "blue"]

# Get unique categories
unique_colors = list(set(colors))

# Create a mapping (color -> index)
color_to_index = {color: idx for idx, color in enumerate(unique_colors)}

# One-hot encoding
one_hot_encoded = []
for color in colors:
    vector = [0] * len(unique_colors)
    vector[color_to_index[color]] = 1
    one_hot_encoded.append(vector)

# Output
print("Unique Colors:", unique_colors)
print("One-Hot Encoded Vectors:")
for vec in one_hot_encoded:
    print(vec)
```
